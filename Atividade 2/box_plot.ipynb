# 1.Boxplot revisada

```python
# Instale uma vez, se precisar: !pip install pandas matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

CAMINHO = "abalone_dataset.csv"  # troque pelo caminho do seu arquivo

df = pd.read_csv(CAMINHO)

colunas_categoricas = [c for c in ["sex", "type"] if c in df.columns]
colunas_numericas = [c for c in df.columns if c not in colunas_categoricas]
df.head()
```

```python
ordem_sexo = ["M", "I", "F"]
tipos = sorted(df["type"].unique())

for col in colunas_numericas:
    fig, axes = plt.subplots(1, len(ordem_sexo), figsize=(14, 4), sharey=True)
    for ax, s in zip(axes, ordem_sexo):
        sns.boxplot(data=df[df["sex"] == s], x="type", y=col, order=tipos,
                    hue="type", hue_order=tipos, legend=False, ax=ax)
        ax.set_title(f"sex = {s}")
        ax.set_xlabel("type")
    axes[0].set_ylabel(col)
    for ax in axes[1:]:
        ax.set_ylabel("")
    fig.suptitle(f"{col} por tipo, separado por sexo", y=1.02)
    plt.tight_layout()
    plt.show()
```
