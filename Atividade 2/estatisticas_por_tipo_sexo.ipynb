# Estatísticas de cada tipo por sexo

1. Quantidade de indivíduos em cada combinação de `type` e `sex`
2. Tabela de médias de todas as colunas
3. Estatísticas completas de cada coluna (contagem, média, desvio padrão, mínimo, quartis, mediana e máximo)
4. Exportar para CSV

```python
# Instale uma vez, se precisar: !pip install pandas
import pandas as pd
from IPython.display import display

pd.set_option("display.float_format", lambda x: f"{x:.4f}")
```

```python
CAMINHO = "abalone_dataset.csv"  # troque pelo caminho do seu arquivo

df = pd.read_csv(CAMINHO)

colunas_numericas = [c for c in df.columns if c not in ["sex", "type"]]
df.head()
```

## 1. Quantidade por tipo e sexo

```python
contagem = pd.crosstab(df["type"], df["sex"], margins=True, margins_name="Total")
contagem
```

## 2. Média de cada coluna por tipo e sexo

```python
medias = df.groupby(["type", "sex"])[colunas_numericas].mean()
medias
```

## 3. Estatísticas completas de cada coluna

Uma tabela por coluna, com uma linha para cada combinação de `type` e `sex`.

```python
estatisticas = {}

for col in colunas_numericas:
    tabela = df.groupby(["type", "sex"])[col].describe()
    tabela = tabela.rename(columns={"50%": "mediana", "25%": "q1", "75%": "q3"})
    estatisticas[col] = tabela
    print(f"=== {col} ===")
    display(tabela)
```

## 4. Exportar para CSV (opcional)

Salva uma tabela única com todas as colunas e estatísticas, para abrir no Excel.

```python
tudo = pd.concat(estatisticas, names=["coluna"]).reset_index()
tudo.to_csv("estatisticas_por_tipo_e_sexo.csv", index=False)
tudo.head()
```
